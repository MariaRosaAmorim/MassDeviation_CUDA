```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("./benchmark_results.csv")
cuda_df = df[df["target"] == "cuda"].sort_values(by="mlups")
c_df = df[df["target"] == "cpu"].sort_values(by="mlups")

plt.plot(
    c_df["grid_size"],
    c_df["mlups"],
    marker="o",
    label="CPU",
)
plt.plot(
    cuda_df["grid_size"],
    cuda_df["mlups"],
    marker="x",
    label="CUDA",
)
plt.xlabel("Grid size")
plt.ylabel("MLUPs")
plt.legend()
plt.show()
```
