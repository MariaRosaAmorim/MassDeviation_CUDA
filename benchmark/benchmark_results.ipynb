```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("./benchmark_results.csv")
cuda_df = df[df["target"] == "cuda"].sort_values(by="mlups")
c_df = df[df["target"] == "cpu"].sort_values(by="mlups")

vmin = df["nt"].min()
vmax = df["nt"].max()

scatter_cpu = plt.scatter(
    c_df["grid_size"] * c_df["nt"],
    c_df["mlups"],
    marker="o",
    label="CPU",
    # c=c_df["nt"],
    # cmap="cool",
    # vmin=vmin,
    # vmax=vmax,
)

scatter_cuda = plt.scatter(
    cuda_df["grid_size"] * cuda_df["nt"],
    cuda_df["mlups"],
    marker="d",
    label="CUDA",
    # c=cuda_df["nt"],
    # cmap="cool",
    # vmin=vmin,
    # vmax=vmax,
)

plt.xlabel("Nodes update count")
plt.ylabel("MLUPs")
plt.legend()

# cbar = plt.colorbar(scatter_cuda)
# cbar.set_label("nt (Number of Time Steps)")

plt.show()
```
